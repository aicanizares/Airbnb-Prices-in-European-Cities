```python
# We will identify variables that are highly correlated to each other.
# 1. rest_index (vertical) and attr_index (horizontal) have a correlation of 0.850229. In an attractive (possibly central) location, there are a lot of restaurants.
# 2. attr_index_norm (vertical) and attr_index (horizontal) have a correlation of 0.714847. The first variable is derived from the second.
# 3. Guest satisfaction and cleanliness rating are strongly correlated (0.714045). Since both influence booking decisions and are crucial for hosts, we’ll keep both.
# 3. Guest satisfaction and leanliness have a strong positive relationship: As one variable increases, the other tends to increase as well.
```

```python
# Conclusions:
# Out of attr_index and attr_index_norm, the one with the lowest correlation to realSum is attr_index at 0.178228. We will then drop attr_index, given also that we keep its derived metric attr_index_norm.
# rest_index and rest_index_norm have a correlation of 0.617970. Out of these 2, rest_index has a lower correlation with the target variable at 0.134722. We will thend drop the column rest_index and keep its derived column, rest_index_norm.
```

```python
# Identifying outliers
# In order to identify outliers, we use the statistical method IQR Method (Tukey's method) if our data does not follow a normal distribution. If it does, we use the Standard Deviation Method.
# In our case, we will use Tukey's method.
```

```python
def tukeys_test_outliers(data):
    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)
    IQR = Q3 - Q1
    
    # Define bounds for the outliers
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Identify the outliers
    outliers = data[(data < lower_bound) | (data > upper_bound)]
    
    return outliers, lower_bound, upper_bound
```

```python
data_series = df_combined['realSum']
outliers, lower_bound, upper_bound = tukeys_test_outliers(data_series)
print(outliers)
```

```python
# Remove outliers
df_no_outliers = df_combined[(df_combined['realSum'] >= lower_bound) & (df_combined['realSum'] <= upper_bound)]
```
