```python
# Create df for weekdays in bcn
df_combined = pd.read_excel("https://github.com/aicanizares/Airbnb-Prices-in-European-Cities/raw/refs/heads/main/df_combined.xlsx")
```

```python
# Do not perform Machine Learning only with the df_weekends, use df_combined.
```

```python
# DO NOT USE ONLY WEEKENDS AS THE FEATURES, USE BOTH WEEKDAYS AND WEEKENDS.
# NOT DONE
```

```python
# Use get_dummies function to perform one-hot encoding (convert categorical variables into multiple binary columns).
df_combined_dummies = pd.get_dummies(df_combined[["room_type", "city"]])
```

```python
df_combined_dummies
```

```python
# Drop "unnamed" column
df_combined = df_combined.drop(columns=['Unnamed: 0'])
```

```python
numeric_cols = df_combined.select_dtypes(include=['number'])
```

```python
numeric_cols.head(5)
```

```python
# Drop columns: latitude and longitude
numeric_cols = numeric_cols.drop(columns=["lng", "lat"])
```

```python
numeric_cols.head(5)
```

```python
# Concatenate features (dfs "numeric_cols" and "df_combined_dummies")
features = pd.concat([numeric_cols, df_combined_dummies], axis=1)
```

```python
features.drop(columns = ["realSum"], inplace=True)
```

```python
features.columns
```

```python
features.columns
```

```python
target = df_combined["realSum"]
```

```python
# Choice between StandardScaler and MinMaxScaler
# We will use MinMaxScaler, because it doesn't assume normality, so it works well with skewed or not normal distributions.
# It rescales values between O and 1 without changing the overall distribution pattern.
```

```python
X_train, X_test, y_train, y_test = train_test_split(features, target, test_size = 0.20, random_state=42)
```

```python
# Perform Grid Search
model = GridSearchCV(estimator=forest, param_grid=grid, cv=5, n_jobs=-1)
```

```python
model.fit(X_train_norm, y_train)
```

```python
# Print Grid Search results
```

```python
print("Best parameters:", model.best_params_)
print("Best R² Score:", model.best_score_)
```

```python
# Train final model with best parameters
best_forest = RandomForestRegressor(**model.best_params_)
best_forest.fit(X_train_norm, y_train)
```

```python
# Print final model score
print("Final Random Forest R² Score:", best_forest.score(X_test_norm, y_test))
```
